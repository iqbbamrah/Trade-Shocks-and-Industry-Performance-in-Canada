# Create Exchange_Rate.csv
- need EX.csv donload from https://data.worldbank.org/indicator/PA.NUS.FCRF?end=2023&locations=CA&start=2013 
- Delete all datea leave exchange reate for Canada only
- Rename that file as EX.csv for further clean up and massageing then create  Exchange_Rate.csv for Data model



```python
# Ref : https://data.worldbank.org/indicator/PA.NUS.FCRF?end=2023&locations=CA&start=2013

import pandas as pd

# Step 1: Load your dataset
Ex = pd.read_csv("../RawData/EX.csv")  

# Step 2: Melt the wide format into long format
ExchangeRate = Ex.melt(id_vars=["Country Name","Country Code"],  
                  var_name="Year", 
                  value_name="CAD_USD_Exchange_Rate" 
                  )

ExchangeRate = ExchangeRate[ExchangeRate["Country Name"] != "United States"]
ExchangeRate = ExchangeRate[ExchangeRate["Country Code"] != "USA"]
ExchangeRate = ExchangeRate.drop(columns= ["Country Name","Country Code"])

# Step 3: Convert Year to integer and Rate to float
ExchangeRate["Year"] = ExchangeRate["Year"].astype(int)
ExchangeRate["CAD_USD_Exchange_Rate"] = pd.to_numeric(ExchangeRate["CAD_USD_Exchange_Rate"], errors="coerce")

# Step 4: Export to CSV
ExchangeRate.to_csv("../Src/Exchange_Rate.csv", index=False)
print(ExchangeRate)
```
