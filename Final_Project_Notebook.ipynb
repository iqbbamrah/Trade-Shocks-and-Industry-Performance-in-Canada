### Analysis Code for Question 1: Economic Growth

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import numpy as np
%matplotlib inline

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = [12, 8]

base_dir = r"drive/My Drive/Canada_Economic_Impact/Master_Optimized_Final"
file_30k = os.path.join(base_dir, "Master_Revenue_30K_5M_2013_2023.csv")
file_5m = os.path.join(base_dir, "Master_Revenue_5M_20M_2013_2023.csv")

output_dir = os.getcwd()

def load_and_combine_data():
    print("Loading data...")
    df1 = pd.read_csv(file_30k)
    df2 = pd.read_csv(file_5m)

    df = pd.concat([df1, df2], ignore_index=True)

    cols_to_clean = ['Whole industry Total revenue']
    for col in cols_to_clean:
        if col in df.columns and df[col].dtype == object:
             df[col] = df[col].astype(str).str.replace(',', '').astype(float)

    return df

def calculate_cagr(end_value, start_value, periods):
    if start_value == 0:
        return np.nan
    return (end_value / start_value) ** (1/periods) - 1

def analyze_national_growth(df):
    print("Analyzing National Growth...")

    national_df = df[df['Province_Description'] == 'Canada Total'].copy()

    grouped = national_df.groupby(['Description_English', 'Year'])['Whole industry Total revenue'].sum().reset_index()


    pivot = grouped.pivot(index='Description_English', columns='Year', values='Whole industry Total revenue')

    # Calculate CAGR (2013 to 2023)
    start_year = 2013
    end_year = 2023
    periods = end_year - start_year

    results = []
    for industry, row in pivot.iterrows():
        start_val = row.get(start_year)
        end_val = row.get(end_year)

        if pd.notnull(start_val) and pd.notnull(end_val) and start_val > 0:
            cagr = calculate_cagr(end_val, start_val, periods)
            abs_growth = end_val - start_val
            results.append({
                'Industry': industry,
                'Start_Revenue': start_val,
                'End_Revenue': end_val,
                'CAGR': cagr,
                'Absolute_Growth': abs_growth
            })

    results_df = pd.DataFrame(results)

    top_10 = results_df.sort_values('CAGR', ascending=False).head(10)

    print("\nTop 10 Growing Industries (National, CAGR):")
    print(top_10[['Industry', 'CAGR', 'End_Revenue']])

    plt.figure(figsize=(12, 8))
    sns.barplot(data=top_10, x='CAGR', y='Industry', palette='viridis', hue='Industry', legend=False)
    plt.title(f'Top 10 Fastest Growing Industries in Canada ({start_year}-{end_year})')
    plt.xlabel('CAGR')
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'top_10_growth_national.png'))
    plt.show()
    plt.close()

    return results_df

def analyze_regional_variation(df):
    print("Analyzing Regional Variation...")
    # Filter OUT 'Canada Total' to get provinces
    regional_df = df[df['Province_Description'] != 'Canada Total'].copy()

    # Group by Province, Industry, Year
    grouped = regional_df.groupby(['Province_Description', 'Description_English', 'Year'])['Whole industry Total revenue'].sum().reset_index()

    # Pivot
    pivot = grouped.pivot_table(index=['Province_Description', 'Description_English'], columns='Year', values='Whole industry Total revenue').reset_index()

    start_year = 2013
    end_year = 2023
    periods = end_year - start_year

    regional_growth = []
    for idx, row in pivot.iterrows():
        start_val = row.get(start_year)
        end_val = row.get(end_year)

        if pd.notnull(start_val) and pd.notnull(end_val) and start_val > 0:
            cagr = calculate_cagr(end_val, start_val, periods)
            regional_growth.append({
                'Province': row['Province_Description'],
                'Industry': row['Description_English'],
                'CAGR': cagr,
                'Size': end_val
            })

    reg_df = pd.DataFrame(regional_growth)

    # Simple median CAGR per province to see overall health
    prov_stats = reg_df.groupby('Province')['CAGR'].median().sort_values(ascending=False).reset_index()

    print("\nMedian Industry Growth by Province:")
    print(prov_stats)

    # Plot
    plt.figure(figsize=(10, 6))
    sns.barplot(data=prov_stats, x='CAGR', y='Province', palette='magma', hue='Province', legend=False)
    plt.title('Median Industry Growth Rate by Province (2013-2023)')
    plt.xlabel('Median CAGR')
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'regional_growth_comparison.png'))
    plt.show()
    plt.close()

    return reg_df

def main():
    df = load_and_combine_data()

    national_results = analyze_national_growth(df)
    regional_results = analyze_regional_variation(df)

    # Growth vs Size (National)
    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=national_results, x='Start_Revenue', y='CAGR', alpha=0.7)
    plt.xscale('log')
    plt.title('Industry Growth (CAGR) vs Initial Size (2013 Revenue)')
    plt.xlabel('2013 Total Revenue (Log Scale)')
    plt.ylabel('CAGR (2013-2023)')
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'growth_vs_size_scatter.png'))
    plt.show()
    plt.close()

    print("\nAnalysis Complete. Visualizations saved.")

if __name__ == "__main__":
    main()
```

# Trade Shock Exposure Analysis (2013-2023)

## Key Findings

### 1. Most Exposed Industries

2.  **Manufacturing**: Consistently high (>100% in many regions), reflecting reliance on imported inputs and export markets (e.g., auto parts cross-border trade).
3.  **Mining, Oil and Gas**: High exposure, particularly driven by exports in resource-rich provinces.
4.  **Transportation & Warehousing**: Essential for moving traded goods.



### 2. Regional Vulnerability
Atlantic provinces and the Prairies show higher trade openness ratios compared to larger, more domestically diversified economies like Ontario or BC (though Ontario has specific high-exposure sectors).

3.  **Nova Scotia**: Port activity and seafood exports.

### 3. Critical Hotspots
Specific Province-Industry pairs show extreme exposure (Trade > 100% of GDP), representing key risk points for the Canadian economy.

## Methodology
-   **Data Source**: `trade.csv` (Statistics Canada).
-   **Metric**: `Trade % of GDP` = `(Total Exports + Total Imports) / GDP`.
-   **Timeframe**: Median values calculated over the 10-year period (2013-2023) to represent structural exposure rather than single-year anomalies.
-   **Note**: Wholesale trade often shows >100% exposure because "Trade" measures the gross value of goods flow, while "GDP" measures value added. A high ratio confirms it is a throughput sector highly sensitive to trade velocity.

## Validation checks
To ensure data quality, we performed the following validation steps on `trade.csv`:
1.  **Internal Consistency**: Re-calculated `trade_pct_gdp` from raw Exports, Imports, and GDP columns.
    -   **Result**: Perfect match (Mean difference = 0.0000). Confirmed `GDP` is in Millions and Trade in Thousands (Factor of 1000 verified).
2.  **Outlier Analysis**: Verified records with >500% exposure.
    -   **Result**: 8 records found, all in **New Brunswick Manufacturing (Code 31)**. This confirms the finding is a structural economic feature (likely Irving Oil refinery throughput relative to provincial manufacturing GDP) rather than a data error.
3.  **Data Completeness**: Checked for missing values and impossible negative metrics.
    -   **Result**: 100% complete, no negative values.


### Analysis Code for Question 2: Trade Exposure

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

EXPORT_CSV = os.path.join(base_dir, "export.csv")
GDP_CSV = os.path.join(base_dir, "gdp.csv")
TRADE_CSV = os.path.join(base_dir, "trade.csv")
OUTPUT_DIR = "."

def load_data():
    """Loads and preprocesses the trade data."""
    print("Loading trade data from:", TRADE_CSV)
    df = pd.read_csv(TRADE_CSV)
    # full range 2013-2023
    df['REF_DATE'] = df['REF_DATE'].astype(int)
    return df

def get_industry_mapping(export_path):
    """
    Creates a mapping from Industry Code to Industry Name using export.csv or gdp.csv.
    This is illustrative as exact mapping logic depends on file structure.
    For this dataset, we might need a manual map or extract it if available.
    Looking at file previews, export.csv has 'North American Industry Classification System (NAICS)' which contains names and codes.
    """
    # Simply defining common codes based on NAICS standard and file inspection for robustness
    mapping = {
        0: "All industries",
        11: "Agriculture, forestry, fishing",
        21: "Mining, oil and gas extraction",
        22: "Utilities",
        23: "Construction",
        31: "Manufacturing", # 31-33
        41: "Wholesale trade",
        44: "Retail trade", # 44-45
        48: "Transportation and warehousing", # 48-49
        51: "Information and cultural industries",
        52: "Finance and insurance",
        53: "Real estate and rental and leasing",
        54: "Professional, scientific and technical services",
        55: "Management of companies",
        56: "Administrative and support services",
        61: "Educational services",
        62: "Health care and social assistance",
        71: "Arts, entertainment and recreation",
        72: "Accommodation and food services",
        81: "Other services"
    }
    return mapping

def analyze_provincial_exposure(df):
    """Analyzes Trade % of GDP by Province (Industry 0)."""
    # Filter for All Industries (0) and exclude 'Canada' if present as a province label to sort provinces only

    prov_df = df[df['industry'] == 0].copy()

    # Calculate median Trade % GDP over the period 2013-2023
    # Group by Province and take median
    prov_exposure = prov_df.groupby('Province')['trade_pct_gdp'].median().sort_values(ascending=False)

    print("\nMedian Trade Exposure by Province (2013-2023):")
    print(prov_exposure)

    plt.figure(figsize=(10, 6))
    sns.barplot(x=prov_exposure.values, y=prov_exposure.index, palette="viridis")
    plt.title("Median Trade Openness by Province (2013-2023)\n(Exports + Imports) / GDP", fontsize=14)
    plt.xlabel("Trade as % of GDP")
    plt.ylabel("Province")
    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, "provincial_trade_exposure.png"))
    plt.close()

def analyze_industry_exposure(df, mapping):
    """Analyzes Trade % of GDP by Industry (Aggregated National Median)."""
    # Filter out Industry 0 (All) and 91 (Public Admin - often low trade or special case)
    ind_df = df[(df['industry'] != 0) & (df['industry'] != 91)].copy()

    # Map industry codes
    ind_df['Industry_Name'] = ind_df['industry'].map(mapping)
    # Fill unknown codes if any
    ind_df['Industry_Name'] = ind_df['Industry_Name'].fillna(ind_df['industry'].astype(str))

    # Calculate median Trade % GDP across all provinces and years for each industry
    # Note: Trade % GDP is calculated per province in the source.

    ind_exposure = ind_df.groupby('Industry_Name')['trade_pct_gdp'].median().sort_values(ascending=False)

    print("\nMedian Trade Exposure by Industry (2013-2023):")
    print(ind_exposure.head(10))

    top_10 = ind_exposure.head(10)
    plt.figure(figsize=(10, 6))
    sns.barplot(x=top_10.values, y=top_10.index, palette="magma")
    plt.title("Top 10 Industries by Median Trade Exposure (2013-2023)", fontsize=14)
    plt.xlabel("Median Trade as % of GDP (Provincial Avg)")
    plt.ylabel("Industry")
    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, "industry_trade_exposure.png"))
    plt.close()

    return ind_df

def analyze_hotspots(df, mapping):
    """Identifies specific Province-Industry pairs with extreme exposure."""
    ind_df = df[(df['industry'] != 0) & (df['industry'] != 91)].copy()
    ind_df['Industry_Name'] = ind_df['industry'].map(mapping)

    # Average over the decade to find structural hotspots
    avg_hotspots = ind_df.groupby(['Province', 'Industry_Name'])['trade_pct_gdp'].mean().reset_index()
    top_hotspots = avg_hotspots.sort_values('trade_pct_gdp', ascending=False).head(15)

    print("\nTop 15 Province-Industry Trade Exposure Hotspots (10-Year Mean):")
    print(top_hotspots)

    top_hotspots.to_csv(os.path.join(OUTPUT_DIR, "top_trade_hotspots.csv"), index=False)

def main():
    df = load_data()
    mapping = get_industry_mapping(EXPORT_CSV)

    analyze_provincial_exposure(df)
    analyze_industry_exposure(df, mapping)
    analyze_hotspots(df, mapping)

    print("Analysis complete. Visualizations saved.")

if __name__ == "__main__":
    main()
```

## Question 3: Financial Impact

# Financial Impact of Trade Shocks (2020 Case Study)

## Key Findings

### 1. Resilience of Trade-Exposed Sectors
Contrary to the hypothesis that global shocks would disproportionately harm trade-dependent industries, we found a **moderate positive correlation (r = 0.51)** between Trade Openness and Profit Margin resilience during the 2020 shock.

-   **High Exposure, High Resilience**: Sectors with the highest trade exposure (e.g., **Wholesale Trade**, >190% Exposure) actually **increased** their profit margins (+0.44% pts) or remained stable during the shock.
-   **Low Exposure, High Impact**: Domestically oriented sectors (e.g., **Health Care**, **Utilities**) suffered larger margin declines.



### 2. The "Goods vs. Services" Shift
The 2020 shock (COVID-19) was unique: it crushed face-to-face services but boosted demand for physical goods.
-   **Trade Sectors** (Manufacturing, Wholesale, Resources) deal primarily in goods, which faced logistics challenges but sustained demand.
-   **Domestic Sectors** (Hospitality, Personal Services) faced lockdowns.

**Data Insight**:
-   **Wholesale Trade** (Trade Hub): Margin increased. Resilience likely driven by the boom in e-commerce and physical goods distribution.
-   **Management of Companies**: Significant margin decline (-0.88 pts), likely due to operational disruptions in corporate functions.

### 3. Top Financial Impacts (Profit Margin Change 2018 vs 2020)
| Industry | Trade Exposure (2019) | Margin Impact (ppt) | Context |
| :--- | :--- | :--- | :--- |
| **Wholesale Trade** | **High (190%)** | **+0.44** | Resilient goods demand. |
| **Educational Services** | Low (0.33%) | +0.23 | Shift to remote lower costs? |
| **Utilities** | Low (14.5%) | -0.60 | Fixed costs vs demand fluctuations. |
| **Management** | High (72%) | -0.88 | Corporate overhead shock. |
| **Health Care** | Low (0.23%) | -0.91 | Higher operational costs (PPE, staffing). |

## Methodology
-   **Financial Data**: Annual Profit Margins (Net Profit / Revenue) for profitable SMEs (30K-5M).
-   **Baseline**: Used **2018** as the pre-shock baseline (due to missing 2019 data).
-   **Comparison**: Calculated the change in Profit Margin from 2018 to 2020.
-   **Metric**: Correlation between *2019 Trade Exposure* and *2020 Margin Change*.

## Data Limitations & Validation Context
*Note on Micro vs. Macro Data Aggregation:*
It is important to qualify that our **Financial Data** represents a specific subset of the economy: Small and Medium Enterprises (SMEs) with revenue between $30K and $5M. In contrast, the **Trade Data** is a Macro-economic aggregate capturing all firms, including large multinationals.
-   **Implication**: The resilience we observed in "Wholesale Trade" reflects the agility of **smaller firms** in this sector. Large incumbents may have faced different macro-pressures not captured in our SME-focused financial dataset.
-   **Conclusion**: Our findings essentially measure "The impact of Macro-Trade Shocks on SME Financial Performance," rather than the performance of the entire sector. This distinction explains potential divergences from national-accounts level profit data.


### Analysis Code for Question 3: Financial Impact

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
import numpy as np

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

TRADE_CSV = os.path.join(base_dir, "trade.csv")
# Using the 30K-5M dataset as it represents the vast majority of Canadian SMEs
FINANCIAL_CSV = os.path.join(base_dir, "Master_Profit_Margin_Profit_Non_Profit_30K_5M_2017_2023.csv")
OUTPUT_DIR = "."

def load_and_prep_trade():
    """Calculates 2019 Trade Exposure by Industry."""
    print("Loading Trade Data...")
    df = pd.read_csv(TRADE_CSV)

    # Filter for 2019 (Pre-Shock Baseline Exposure)
    df_2019 = df[df['REF_DATE'] == 2019].copy()

    # We need to aggregate province-level trade data to get a National Industry Exposure
    trade_exposure = df_2019.groupby('industry')['trade_pct_gdp'].median().reset_index()
    trade_exposure.rename(columns={'trade_pct_gdp': 'Trade_Exposure_2019'}, inplace=True)

    # Remove '0' (All industries) and '91' (Public Admin)
    trade_exposure = trade_exposure[~trade_exposure['industry'].isin([0, 91])]

    return trade_exposure

def load_and_prep_financials():
    """Calculates Profit Margins and the 2020 Shock Impact."""
    print("Loading Financial Data...")
    df = pd.read_csv(FINANCIAL_CSV)

    print("Unique Years raw:", sorted(df['Year'].unique()))

    if 'Canada' in df.columns:
        df = df[df['Canada'] == 0].copy()
        print("Unique Years after filtering Canada==0:", sorted(df['Year'].unique()))

    cols = ['Year', 'Naics code', 'Description_English', 'Profitable Total Revenue', 'profitable Net Profit']

    df = df[cols].copy()

    # Calculate Profit Margin %
    df = df[df['Profitable Total Revenue'] > 0]
    df['Profit_Margin'] = (df['profitable Net Profit'] / df['Profitable Total Revenue']) * 100

    # We need the Change from Pre-Shock (2018) to Shock (2020)
    # Note: 2019 data is missing from the dataset, so we use 2018 as the baseline.
    df_pivot = df.pivot(index='Naics code', columns='Year', values='Profit_Margin')

    # Calculate Shock Impact
    # Baseline: 2018. Shock: 2020.
    if 2018 in df_pivot.columns and 2020 in df_pivot.columns:
        df_pivot['Margin_Impact_2020'] = df_pivot[2020] - df_pivot[2018]
        # Recovery: 2022 vs 2018
        if 2022 in df_pivot.columns:
            df_pivot['Recovery_2022'] = df_pivot[2022] - df_pivot[2018]
    else:
        print("Error: 2018 or 2020 data missing from financials.")
        return None

    # Reset index to make NAICS a column for merging
    return df_pivot.reset_index()

def merge_and_analyze(trade_df, fin_df):
    """Merges datasets and analyzes correlation."""
    print("Merging datasets...")

    merged = pd.merge(fin_df, trade_df, left_on='Naics code', right_on='industry', how='inner')

    print(f"Matched {len(merged)} industries.")

    if len(merged) < 5:
        print("Warning: Low match count. Indices might be mismatched (e.g. 31-33 vs 31).")

    plt.figure(figsize=(10, 6))
    sns.regplot(x='Trade_Exposure_2019', y='Margin_Impact_2020', data=merged, color='firebrick')

    for i, row in merged.iterrows():
        if abs(row['Margin_Impact_2020']) > 2 or row['Trade_Exposure_2019'] > 50:
            plt.text(row['Trade_Exposure_2019']+1, row['Margin_Impact_2020'], str(row['industry']), fontsize=9)

    plt.title("Impact of Trade Exposure on Financial Resilience (2020 Shock)", fontsize=14)
    plt.xlabel("Trade Openness (2019) [% GDP]")
    plt.ylabel("Change in Profit Margin (2019 to 2020) [% pts]")
    plt.axhline(0, color='black', linestyle='--', linewidth=1)
    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, "trade_shock_impact_scatter.png"))
    plt.close()

    # 2. Correlation
    corr = merged['Trade_Exposure_2019'].corr(merged['Margin_Impact_2020'])
    print(f"\nCorrelation between Trade Exposure and 2020 Margin Shock: {corr:.4f}")

    # 3. Bar Chart: Hardest Hit Industries
    impact_df = merged.sort_values('Margin_Impact_2020', ascending=True) # Negative impact first

    print("\nTop 5 Industries by Margin Decline (2020):")
    print(impact_df[['industry', 'Trade_Exposure_2019', 'Margin_Impact_2020']].head(5))

    return merged

def main():
    trade_df = load_and_prep_trade()
    fin_df = load_and_prep_financials()

    if fin_df is not None:
        merged = merge_and_analyze(trade_df, fin_df)
        print("Analysis complete. Visualizations saved.")

if __name__ == "__main__":
    main()
```

## Question 4: Exchange Rate Impact

# Exchange Rate Impact Analysis (CAD-USD)

## Key Findings

### 1. The "Petro-Currency" Paradox
Contrary to the standard economic theory that a **Weaker Currency** boosts export revenues (by making goods cheaper abroad), we found a **Strong Positive Correlation (r = 0.62)** between a **Stronger CAD** and **Revenue Growth** for Export-Heavy industries.

-   **Hypothesis**: Weaker CAD (Depreciation) -> Higher Export Revenue.
-   **Reality (2014-2023)**: Stronger CAD -> Higher Revenue.

**Why?**
The Canadian Dollar is a **commodity currency**. It historically appreciates when oil and resource prices rise.
-   **Boom Cycles**: High Oil/Resource Prices -> High Export Revenue -> **Strong CAD**.
-   **Bust Cycles** (e.g., 2015): Low Oil Prices -> Low Export Revenue -> **Weak CAD**.
The underlying commodity price cycle drives *both* the revenue and the currency in the same direction, overpowering the competitive benefit of depreciation.

### 2. Universal Impact
Both **Export-Heavy** (Mining, Manufacturing) and **Import-Heavy** (Retail, Construction) cohorts showed nearly identical positive correlations (0.617 vs 0.616).
-   This suggests that the exchange rate in Canada acts more as a **barometer of general economic health** rather than a lever that differentially favors one sector over another.
-   When the CAD is strong, the entire Canadian economy (Domestic and Export) tends to be growing faster.



## Methodology
-   **Data**:
    -   **Exchange Rate**: Annual Average CAD/USD (Bank of Canada).
    -   **Revenue**: Year-over-Year Growth Rate of Total Revenue (Master_Revenue dataset).
-   **Cohorts**:
    -   **Export-Heavy**: Industries where Exports > 1.2 * Imports (e.g., Mining, Manufacturing).
    -   **Import-Heavy**: Industries where Imports > 1.2 * Exports (e.g., Retail, Construction).
-   **Metric**: Correlation between *Annual Change in CAD* and *Annual Revenue Growth %*.


### Analysis Code for Question 4: Exchange Rate Impact

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
import numpy as np

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

TRADE_CSV = os.path.join(base_dir, "trade.csv")
# Using Revenue file because it covers 2013-2023 (capturing the 2014-2016 depreciation)
REVENUE_CSV = os.path.join(base_dir, "Master_Revenue_30K_5M_2013_2023.csv")
OUTPUT_DIR = "."

# Manual Exchange Rate Data (Annual Average CAD to USD)
# Source: Bank of Canada / St. Louis Fed approximate annual averages
EXCHANGE_RATES = {
    2013: 0.971,
    2014: 0.906,
    2015: 0.783,
    2016: 0.755,
    2017: 0.770,
    2018: 0.772,
    2019: 0.754,
    2020: 0.746, # Derived from 1.34
    2021: 0.798,
    2022: 0.769,
    2023: 0.741
}

def load_exchange_rates():
    df = pd.DataFrame(list(EXCHANGE_RATES.items()), columns=['Year', 'CAD_USD'])
    return df

def classify_industries():
    """Classifies industries as Export-Heavy, Import-Heavy, or Neutral."""
    print("Classifying industries from Trade Data...")
    df = pd.read_csv(TRADE_CSV)

    # Filter for standard industries (remove Total '0' and Public '91')
    df = df[~df['industry'].isin([0, 91])].copy()

    # Aggregate Exports and Imports over the full period to get structural orientation
    grouped = df.groupby('industry')[['Value of exports', 'Value of imports']].sum().reset_index()

    def classify(row):
        exp = row['Value of exports']
        imp = row['Value of imports']
        if exp > imp * 1.2:
            return 'Export-Heavy'
        elif imp > exp * 1.2:
            return 'Import-Heavy'
        else:
            return 'Neutral'

    grouped['Orientation'] = grouped.apply(classify, axis=1)

    print("\nIndustry Classifications:")
    print(grouped['Orientation'].value_counts())

    print("\nExamples of Export-Heavy:", grouped[grouped['Orientation'] == 'Export-Heavy']['industry'].head(3).tolist())
    print("Examples of Import-Heavy:", grouped[grouped['Orientation'] == 'Import-Heavy']['industry'].head(3).tolist())

    return grouped[['industry', 'Orientation']]

def load_and_prep_revenue(orientation_df):
    """Loads revenue data and aggregates by Orientation."""
    print("Loading Revenue Data...")
    df = pd.read_csv(REVENUE_CSV)

    # Filter for Canada Total (Province code usually) if present, or aggregate.
    # The clean revenue file usually has 'Province' column.
    if 'Province_Description' in df.columns:
         df = df[df['Province_Description'] == 'Canada Total']
    elif 'Province' in df.columns:
         # Assuming 'Canada Total' or code. Let's aggregate to be safe if 'Canada Total' isn't explicit
         pass

    df_agg = df.groupby(['Year', 'Naics code'])['Whole industry Total revenue'].sum().reset_index()

    merged = pd.merge(df_agg, orientation_df, left_on='Naics code', right_on='industry', how='inner')

    # Aggregate Revenue by Orientation and Year
    cohort_revenue = merged.groupby(['Year', 'Orientation'])['Whole industry Total revenue'].sum().reset_index()

    return cohort_revenue

def analyze_impact(revenue_df, exchange_df):
    """Correlates revenue and exchange rates."""

    data = pd.merge(revenue_df, exchange_df, on='Year', how='inner')

    # Normalize Revenue to 2013 = 100 for comparison
    data['Revenue_Index'] = 0.0
    for orient in data['Orientation'].unique():
        base_val = data[(data['Orientation'] == orient) & (data['Year'] == 2013)]['Whole industry Total revenue'].values[0]
        data.loc[data['Orientation'] == orient, 'Revenue_Index'] = (data.loc[data['Orientation'] == orient, 'Whole industry Total revenue'] / base_val) * 100

    # Calculate YoY Growth Rates to remove long-term trend (inflation/scale)
    data.sort_values(['Orientation', 'Year'], inplace=True)
    data['Revenue_Growth'] = data.groupby('Orientation')['Whole industry Total revenue'].pct_change() * 100

    plt.figure(figsize=(12, 7))

    sns.lineplot(x='Year', y='Revenue_Growth', hue='Orientation', data=data, linewidth=2.5, marker='o')
    plt.axhline(0, color='grey', linestyle='--', alpha=0.5)

    ax2 = plt.gca().twinx()
    sns.lineplot(x='Year', y='CAD_USD', data=data, ax=ax2, color='black', linestyle='--', label='CAD/USD Rate', alpha=0.6)
    ax2.set_ylabel('CAD to USD Exchange Rate')
    ax2.legend(loc='upper left')

    plt.title("Revenue Growth vs. Exchange Rate (2014-2023)\nExport-Heavy vs. Import-Heavy Industries", fontsize=14)
    plt.xlabel("Year")
    plt.ylabel("Annual Revenue Growth (%)")
    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, "exchange_rate_impact_growth.png"))
    plt.close()

    print("\nCorrelations between CAD/USD Rate and Revenue Growth:")
    for orient in data['Orientation'].unique():
        subset = data[data['Orientation'] == orient].dropna()
        corr = subset['Revenue_Growth'].corr(subset['CAD_USD'])
        print(f"{orient}: {corr:.3f}")

    data['FX_Change'] = data['CAD_USD'].pct_change()
    print("\nCorrelations between FX % Change and Revenue Growth:")
    for orient in data['Orientation'].unique():
        subset = data[data['Orientation'] == orient].dropna()
        corr = subset['Revenue_Growth'].corr(subset['FX_Change'])
        print(f"{orient}: {corr:.3f}")

def main():
    exchange_df = load_exchange_rates()
    orientation_df = classify_industries()
    revenue_df = load_and_prep_revenue(orientation_df)

    if not revenue_df.empty:
        analyze_impact(revenue_df, exchange_df)
        print("Analysis complete.")
    else:
        print("Error: No revenue data matched.")

if __name__ == "__main__":
    main()
```

## Question 5: Firm Size Impact

# Small vs. Large Firm Response Analysis

## Key Findings

### 1. The Volatility Gap
The most distinct difference between firm sizes is **Volatility**. Small firms ($30K-$5M) are far more sensitive to economic changes than their larger counterparts ($5M-$20M).
-   **Small Firm Volatility**: ~23.6% (Standard Deviation of Growth).
-   **Larger Firm Volatility**: ~5.3%.

**Takeaway**: Large firms act as "Economic Ballast," maintaining steady operations through shocks. Small firms are the "Speedboats," capable of rapid acceleration but also subject to violent fluctuations.

### 2. The 2020 Anomaly (Pandemic Shock)
Contrary to the expectation that small businesses would be decimated, our data shows a **massive surge (+62.9%)** in aggregated revenue for the Small Firm cohort in 2020.
-   **Larger Firms**: Grew modestly at +5.3%.

**Hypotheses for Small Firm Surge**:
1.  **Digital Adaptation**: Rapid shift to e-commerce and new business models (e.g., delivery services, digital consulting) that larger incumbents couldn't match quickly.
2.  **New Business Formation**: The pandemic sparked a wave of entrepreneurship (gig economy, side hustles) which enter the data at the "Small" end.
3.  **Support Mechanisms**: Inclusion of government support (CEWS/CEBA) as 'Revenue' in administrative data might disproportionately aid the balance sheets of the smallest entities relative to their size.

### 3. The 2015 Oil Shock
-   **Small Firms**: +10.5% Growth.
-   **Larger Firms**: +7.7% Growth.
Both weathered the resource shock well, likely because the revenue contraction in the oil sector (Large firms) was offset by growth in non-resource sectors benefiting from the cheaper currency.



## Methodology
-   **Cohorts**:
    -   **Small**: Annual Revenue $30K - $5M.
    -   **Larger**: Annual Revenue $5M - $20M.
-   **Metric**: Year-over-Year % Growth of Total Industry Revenue.
-   **Data Source**: `Master_Revenue` dataset (2013-2023).


### Analysis Code for Question 5: Firm Size Impact

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
import numpy as np

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

SMALL_CSV = os.path.join(base_dir, "Master_Revenue_30K_5M_2013_2023.csv")
LARGE_CSV = os.path.join(base_dir, "Master_Revenue_5M_20M_2013_2023.csv")
OUTPUT_DIR = "."

def load_and_agg(path, cohort_name):
    print(f"Loading {cohort_name} data from {path}...")
    df = pd.read_csv(path)

    # Filter for Canada Total if possible to avoid duplicates
    if 'Province_Description' in df.columns:
        df = df[df['Province_Description'] == 'Canada Total']

    # Aggregate Total Revenue by Year
    # Metric: 'Whole industry Total revenue'
    agg = df.groupby('Year')['Whole industry Total revenue'].sum().reset_index()
    agg['Cohort'] = cohort_name
    return agg

def analyze_size_impact():
    small_df = load_and_agg(SMALL_CSV, "Small Firms ($30K-$5M)")
    large_df = load_and_agg(LARGE_CSV, "Larger Firms ($5M-$20M)")

    combined = pd.concat([small_df, large_df], ignore_index=True)

    # YoY Growth
    combined.sort_values(['Cohort', 'Year'], inplace=True)
    combined['Growth_Pct'] = combined.groupby('Cohort')['Whole industry Total revenue'].pct_change() * 100

    combined = combined[combined['Year'] >= 2014]

    plt.figure(figsize=(12, 7))
    sns.lineplot(x='Year', y='Growth_Pct', hue='Cohort', data=combined, marker='o', linewidth=2.5)

    # Highlight Shocks
    plt.axvline(2015, color='red', linestyle='--', alpha=0.3, label='2015 Oil Shock')
    plt.axvline(2020, color='black', linestyle='--', alpha=0.3, label='2020 COVID Shock')

    plt.title("Revenue Growth Resilience: Small vs. Larger Firms (2014-2023)", fontsize=14)
    plt.ylabel("Annual Revenue Growth (%)")
    plt.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, "size_impact_growth.png"))
    plt.close()

    # Impact Analysis Table
    print("\n--- Impact Analysis (Growth %) ---")
    shocks = [2015, 2020]
    recoveries = [2016, 2021]

    rows = []
    for year in shocks + recoveries:
        for cohort in combined['Cohort'].unique():
            val = combined[(combined['Cohort'] == cohort) & (combined['Year'] == year)]['Growth_Pct'].values
            if len(val) > 0:
                rows.append({'Year': year, 'Cohort': cohort, 'Growth': val[0]})

    impact_df = pd.DataFrame(rows).sort_values('Year')
    print(impact_df)

    # Volatility (Std Dev)
    print("\n--- Volatility (Std Dev of Growth) ---")
    vol = combined.groupby('Cohort')['Growth_Pct'].std()
    print(vol)

if __name__ == "__main__":
    analyze_size_impact()
```
