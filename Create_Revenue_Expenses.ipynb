# Revenue file

```python
# Revenue file
# According to Data dictionary revenue has 3 cateagory data.

import pandas as pd

revenue30K = pd.read_csv("../RawData/Master_Optimized_Final/Master_Revenue_30K_5M_2013_2023.csv")
revenue5M = pd.read_csv("../RawData/Master_Optimized_Final/Master_Revenue_5M_20M_2013_2023.csv")

revenue30K["file_name"]= "revenue30K"
revenue5M["file_name"]= "revenue5M"


revenue30K.info()
revenue5M.info()

Revenue=pd.concat([revenue30K,revenue5M], ignore_index=True)

Revenue[["Year", "Incorporation Status","Province","Naics code"]] = Revenue[["Year", "Incorporation Status","Province","Naics code"]].astype(int)

Revenue.rename(columns={"Naics code": "NAICS", "Year": "Year"}, inplace=True)

#  Keep only status 2 Incorporate data
# Revenue = Revenue[Revenue["Incorporation Status"] == 2]

# print(Revenue.head())

print("revenue30K : ",len(revenue30K))
print("revenue5M : " ,len(revenue5M))
print("Revenue : " , len(Revenue))

Revenue.to_csv("../RawData/Revenue.csv",index = False)
```

# Expenses file

```python
import pandas as pd

# Expenses file
expenses30K = pd.read_csv("../RawData/Master_Optimized_Final/Master_Expenses_Value_30K_5M_2013_2023.csv")
expenses5M = pd.read_csv("../RawData/Master_Optimized_Final/Master_Expenses_Value_5M_20M_2013_2023.csv")
   
expenses30K["file_name"]= "expenses30K"
expenses5M["file_name"]= "expenses5M"

Expenses=pd.concat([expenses30K,expenses5M], ignore_index=True)

Expenses[["Year","Naics code"]] = Expenses[["Year","Naics code"]].astype(int)


Expenses.rename(columns={"Naics code": "NAICS", "Year": "Year"}, inplace=True)

# print(Revenue.head())

print("expenses30K : ",len(expenses30K))
print("expenses5M : " ,len(expenses5M))
print("Expenses : " , len(Expenses))

Expenses.to_csv("../RawData/Expenses.csv",index = False)
```
