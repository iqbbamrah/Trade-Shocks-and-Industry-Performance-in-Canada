```python

import pandas as pd
gdp = pd.read_csv("../RawData/gdp.csv")  
```

```python
gdp["Geography"] = gdp["Geography"].fillna(method="ffill")
gdp["North American Industry Classification System (NAICS)"] = gdp["North American Industry Classification System (NAICS)"].fillna(method="ffill")

gdp.head(30)
```

```python
gdp.info()
```

```python
gdp["industry"] = gdp["North American Industry Classification System (NAICS)"].str.extract("(\d+)")
gdp["industry"] = gdp["industry"].fillna(0).astype(int)
gdp["industry"] = gdp["industry"].astype(int)
gdp.head()
```

```python
gdp = gdp.drop(columns=["North American Industry Classification System (NAICS)"])
gdp["GDP"] = gdp["GDP"].str.replace(",", "").astype(float).astype(int)
gdp.head()
```

```python
gdp = gdp.rename(columns={"REF-Date": "REF_DATE"})
gdp = gdp.rename(columns={"Geography": "Province"})
gdp.head()
```

```python
gdp["industry"] = gdp["industry"].replace(1, 0)
gdp.info()
```

```python
gdp.to_csv("../Src/TradetoGDP.csv", index=False)
print("Trade csv created in folder ../Src/TradetoGDP.csv")
```
